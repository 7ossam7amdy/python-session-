```python
nEven, nOdd = 0, 0
numper = input("Enter a number: ")

while numper != "0":
    if int(numper) % 2 == 0:
        nEven += 1
    else:
        nOdd += 1
    numper = input("Enter a number: ")  
print("Even numbers:", nEven)
print("Odd numbers:", nOdd)
```

```python
blocks = int(input("Enter the number of blocks: "))

height = 0
in_layer = 1

while in_layer <= blocks:
    height += 1
    blocks -= in_layer
    in_layer += 1

print("The height of the pyramid:", height)
```
